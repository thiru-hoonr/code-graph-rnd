# False positives, round 2: syntax rules (kotlin, gdscript, nix, verilog, rescript, solidity, elixir) on CRG's output
Rules in `notebooks/fp_round2.py`. Each looks only at the source and CRG's own rows, never at the reference.
Before = CRG as stored. After = CRG with the rule applied. Same reference, same matching (`survey_lab.py`).

```python
import survey_lab as L, fp_round2 as F, pandas as pd, copy
rows = []
for lang, fn in F.APPLY.items():
    info = L.repos()[lang]; repo = L.ROOT / 'apps/survey' / info['folder']
    files = L.files_for(repo, info['ext'])
    R, NA, errors, ref_label, _ = L.reference(lang, repo, files)
    G = L.crg(info['db'], repo, files, lang)
    before, _ = L.score(R, G, NA)
    after, _ = L.score(R, fn(copy.deepcopy(G), repo), NA)
    for k in before.index:
        if before.loc[k, 'note']:
            continue
        rows.append({'language': lang, 'rule': F.RULES[lang], 'kind': k,
                     'CRG extra before': before.loc[k, 'CRG extra'], 'CRG extra after': after.loc[k, 'CRG extra'],
                     'precision before %': before.loc[k, 'precision %'], 'precision after %': after.loc[k, 'precision %'],
                     'recall before %': before.loc[k, 'recall %'], 'recall after %': after.loc[k, 'recall %'],
                     'expected': before.loc[k, 'expected'],
                     'CRG reported before': before.loc[k, 'CRG'], 'CRG reported after': after.loc[k, 'CRG'],
                     'matched before': before.loc[k, 'matched'], 'matched after': after.loc[k, 'matched']})
T = pd.DataFrame(rows)
T = T[(T['CRG extra before'] != T['CRG extra after']) | (T['recall before %'] != T['recall after %'])]
T.to_csv(L.ROOT / 'results/language-survey/fp_round2.csv', index=False)
T
```
